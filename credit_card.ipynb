<span style="color:brown; font-size:18px">This code splits your data into input (features) and output (target) to get ready for making predictions</span>

```python
X = df.drop(columns=['Class'])
Y = df['Class']

print("Features shape:", X.shape)
print("Target shape:", Y.shape)

# Convert to NumPy arrays for easier processing
features_array = X.values
target_array = Y.values
```

<span style="color:brown; font-size:18px">This code splits your data into two groups to one group to teach the computer (training) and one group to check if it learned well (testing). 
It uses 80% for training and 20% for testing.</span>

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

<span style="color:brown; font-size:18px">
This code uses the Random Forest algorithm to train a model and make predictions.  
It first trains the model on training data (`x_train`, `y_train`),  
and then predicts the results for test data (`x_test`).  
The predictions are stored in `y_pred`.

Random Forest is an `ensemble method`, which means it combines the results of `many decision trees` 
to make better and more accurate predictions.  
</span>

```python
from sklearn.ensemble import RandomForestClassifier
random = RandomForestClassifier()

random.fit(x_train, y_train)
y_pred = random.predict(x_test)
```

<span style="color:brown; font-size:18px">
This code checks for missing values (NaNs) in `y_test` and fills them using the **most frequent value**.

Steps:
1. It first prints how many NaN values are in `y_test`.
2. Then it uses `SimpleImputer` from `sklearn` with the strategy `'most_frequent'` to fill in those NaNs.
3. Since `SimpleImputer` needs 2D input, `y_test` is reshaped after converting it to a NumPy array.
4. After filling, the result is flattened back into a 1D array.
5. Finally, it prints the number of NaNs after imputation (should be 0).

This is useful for cleaning missing labels before using them in a machine learning model.
</span>

```python
from sklearn.impute import SimpleImputer
print("NaN values in y_test before imputation:", np.isnan(y_test).sum())

imputer = SimpleImputer(strategy='most_frequent')

y_test = imputer.fit_transform(y_test.reshape(-1, 1)).flatten()
print("NaN values in y_test after imputation:", np.isnan(y_test).sum())
```

<span style="color:brown; font-size:18px">

This code checks how well the Random Forest model works by calculating important scores like accuracy, precision, recall, F1 score, and MCC.  
It also shows a confusion matrix and a bar chart to make the results easy to understand.

</span>


```python
from sklearn.metrics import (
    classification_report, accuracy_score, precision_score, 
    recall_score, f1_score, matthews_corrcoef, confusion_matrix
)

n_outliers = len(fraud)

# Count the number of errors in predictions
n_errors = (y_pred != y_test).sum()

print("Model Evaluation Report - Random Forest Classifier")

# Accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f}")

# Precision
precision = precision_score(y_test, y_pred)
print(f"Precision: {precision:.4f}")

# Recall
recall = recall_score(y_test, y_pred)
print(f"Recall: {recall:.4f}")

# F1 Score 
f1 = f1_score(y_test, y_pred)
print(f"F1 Score: {f1:.4f}")

# MCC
mcc = matthews_corrcoef(y_test, y_pred)
print(f"Matthews Correlation Coefficient (MCC): {mcc:.4f}")

# confusion matrix
cm = confusion_matrix(y_test, y_pred)
print("\nConfusion Matrix:")
print(cm)

# full classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
labels = ['Not Fraud', 'Fraud']

# Plot confusion matrix 
plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.title('Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.tight_layout()
plt.show()

# Prepare metrics for bar chart
metrics = {
    'Accuracy': accuracy,
    'Precision': precision,
    'Recall': recall,
    'F1 Score': f1,
    'MCC': mcc
}

# Plot bar chart of metrics
plt.figure(figsize=(8, 5))
sns.barplot(x=list(metrics.keys()), y=list(metrics.values()), palette='Set2')
plt.ylim(0, 1)
plt.title('Evaluation Metrics')
plt.ylabel('Score')

# Show value labels on bars
for i, v in enumerate(metrics.values()):
    plt.text(i, v + 0.02, f"{v:.2f}", ha='center', fontsize=10)

plt.tight_layout()
plt.show()
```
